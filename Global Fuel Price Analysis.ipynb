## 8.Fuel price visualization and analysis

This line plot shows the average global gasoline price over time.

```python
plt.figure(figsize=(10,5))
df.groupby('date')['gasoline_usd_per_liter'].mean().plot(color="red")
plt.title("Global  Gasoline Price Trend")
plt.xlabel("Year")
plt.ylabel("USD per Liter")
plt.show()
```

According to this line plot year 2020 has a globaly lowest gasoline price and year 2022 has a globaly highest gasoline price

```python
plt.figure(figsize=(7,5))
plt.scatter(df['brent_crude_usd_bbl'], df['gasoline_usd_per_liter'],color="orange")
plt.xlabel("Brent Crude Price")
plt.ylabel("Gasoline Price")
plt.title("Crude Oil vs Gasoline Price")
plt.show()
```

This scatter plot shows the relationship between Brent crude oil prices and gasoline prices.so we can see that when crude price increases then gasoline price increases

```python
country_avg = df.groupby('country')['gasoline_usd_per_liter'].mean().sort_values()
country_avg.plot(kind='bar')
plt.title("Gasoline Price by Country")
plt.show()
```

This bar chart shows the average gasoline price for each country.

```python
region_avg = df.groupby('region')['gasoline_usd_per_liter'].mean()

region_avg.plot(kind='bar', figsize=(8,5))
plt.style.use('ggplot')
plt.title("Average Fuel Price by Region")
plt.show()
```

This bar chart shows the average gasoline price across different regions.


```python
plt.figure(figsize=(8,5))
plt.hist(df['gasoline_usd_per_liter'], bins=10 ,color='green')
plt.title("Distribution of Gasoline Prices")
plt.xlabel("Gasoline Price (USD/Liter)")
plt.ylabel("Frequency")
plt.show()
```

This histogram shows the distribution of gasoline prices.


```python
plt.figure(figsize=(12,6))
df.boxplot(column='gasoline_usd_per_liter', by='country')
plt.xticks(rotation=45)
plt.title("Fuel Price Distribution by Country")
plt.suptitle("")
plt.show()
```

The boxplot shows all country have a outliers it shows that gasoline prices are not stable in any country

```python
top5 = country_avg.sort_values(ascending=False).head(5)
plt.figure()
plt.bar(top5.index, top5.values,color=['red', 'blue', 'green', 'orange', 'purple'])
plt.title("Top 5 Highest Fuel Price Countries")
plt.xticks(rotation=45)
plt.show()
```

This bar chart shows the top 5 countries with the highest average gasoline prices.
